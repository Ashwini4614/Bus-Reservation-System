```python
### ==========================================
# STEP 1: BUS DATA
# ==========================================

buses = {
    "B101": {
        "route": "Pune -> Mumbai",
        "departure": "08:00 AM",
        "arrival": "12:00 PM",
        "total_seats": 10,
        "fare": 450,
        "booked_seats": []
    },

    "B102": {
        "route": "Pune -> Nashik",
        "departure": "09:30 AM",
        "arrival": "01:00 PM",
        "total_seats": 12,
        "fare": 350,
        "booked_seats": []
    }
}


# ==========================================
# BOOKING STORAGE
# ==========================================

bookings = {}

# Starting number for booking ID
booking_counter = 1000


# ==========================================
# VIEW BUSES
# ==========================================

def view_buses():

    print("\n")
    print("=" * 75)
    print("                    AVAILABLE BUSES")
    print("=" * 75)

    print(
        f"{'ID':<8}"
        f"{'ROUTE':<25}"
        f"{'DEPARTURE':<15}"
        f"{'ARRIVAL':<15}"
        f"{'FARE':<10}"
        f"{'SEATS'}"
    )

    print("-" * 75)

    for bus_id, bus in buses.items():

        available_seats = (
            bus["total_seats"] - len(bus["booked_seats"])
        )

        print(
            f"{bus_id:<8}"
            f"{bus['route']:<25}"
            f"{bus['departure']:<15}"
            f"{bus['arrival']:<15}"
            f"₹{bus['fare']:<9}"
            f"{available_seats}"
        )

    print("=" * 75)


# ==========================================
# VIEW AVAILABLE SEATS
# ==========================================

def view_seats():

    bus_id = input("\nEnter Bus ID: ").strip().upper()

    if bus_id not in buses:
        print("Invalid Bus ID.")
        return

    bus = buses[bus_id]

    print("\n" + "=" * 40)
    print(f"             BUS {bus_id}")
    print("=" * 40)

    for seat in range(1, bus["total_seats"] + 1):

        if seat in bus["booked_seats"]:
            print(f"[{seat:02d}] - Booked")
        else:
            print(f"[{seat:02d}] - Available")

    available_seats = [
        seat
        for seat in range(1, bus["total_seats"] + 1)
        if seat not in bus["booked_seats"]
    ]

    print("\nAvailable Seats:", available_seats)

    print("=" * 40)


# ==========================================
# CALCULATE FARE
# ==========================================

def calculate_fare(base_fare, age):

    if age < 12:
        # 50% discount
        fare = base_fare * 0.50

    elif age >= 60:
        # 30% discount
        fare = base_fare * 0.70

    else:
        # Full fare
        fare = base_fare

    return fare


# ==========================================
# GENERATE UNIQUE BOOKING ID
# ==========================================

def generate_booking_id():

    global booking_counter

    booking_counter += 1

    booking_id = "ST" + str(booking_counter)

    return booking_id


# ==========================================
# STEP 4: BOOK TICKET
# ==========================================

def book_ticket():

    print("\n")
    print("=" * 40)
    print("             BOOK TICKET")
    print("=" * 40)

    # --------------------------------------
    # 1. Passenger Name
    # --------------------------------------

    passenger_name = input("Enter passenger name: ").strip().title()

    if passenger_name == "":
        print("Passenger name cannot be empty.")
        return

    # --------------------------------------
    # 2. Passenger Age
    # --------------------------------------

    try:

        age = int(input("Enter age: "))

    except ValueError:

        print("Invalid input.")
        print("Age must be a number.")
        return

    if age <= 0:

        print("Age must be greater than 0.")
        return

    # --------------------------------------
    # 3. Select Bus
    # --------------------------------------

    print("\nAvailable Buses:")

    for bus_id, bus in buses.items():

        available = (
            bus["total_seats"] -
            len(bus["booked_seats"])
        )

        print(
            f"{bus_id} - "
            f"{bus['route']} - "
            f"₹{bus['fare']} - "
            f"{available} seats available"
        )

    bus_id = input("\nEnter Bus ID: ").strip().upper()

    # Check Bus ID
    if bus_id not in buses:

        print("Invalid Bus ID.")
        return

    bus = buses[bus_id]

    # --------------------------------------
    # 4. Check Available Seats
    # --------------------------------------

    if len(bus["booked_seats"]) == bus["total_seats"]:

        print("Sorry, this bus is fully booked.")
        return

    available_seats = [
        seat
        for seat in range(1, bus["total_seats"] + 1)
        if seat not in bus["booked_seats"]
    ]

    print("\nAvailable seats:", available_seats)

    # --------------------------------------
    # 5. Select Seat
    # --------------------------------------

    try:

        seat = int(input("Select seat number: "))

    except ValueError:

        print("Invalid input.")
        print("Seat number must be a number.")
        return

    # Check seat range
    if seat < 1 or seat > bus["total_seats"]:

        print("Invalid seat number.")
        return

    # Check whether seat is already booked
    if seat in bus["booked_seats"]:

        print("Sorry, this seat is already booked.")
        return

    # --------------------------------------
    # 6. Calculate Fare
    # --------------------------------------

    fare = calculate_fare(bus["fare"], age)

    # --------------------------------------
    # 7. Generate Booking ID
    # --------------------------------------

    booking_id = generate_booking_id()

    # --------------------------------------
    # 8. Book the Seat
    # --------------------------------------

    bus["booked_seats"].append(seat)

    # --------------------------------------
    # 9. Store Booking Information
    # --------------------------------------

    bookings[booking_id] = {

        "passenger_name": passenger_name,

        "age": age,

        "bus_id": bus_id,

        "route": bus["route"],

        "seat": seat,

        "fare": fare,

        "status": "CONFIRMED"
    }

    # --------------------------------------
    # 10. Display Confirmation
    # --------------------------------------

    print("\n")
    print("=" * 45)
    print("           BOOKING CONFIRMED")
    print("=" * 45)

    print("Booking ID :", booking_id)
    print("Passenger_Name  :", passenger_name)
    print("Age        :", age)
    print("Journey    :", bus_id)
    print("Route      :", bus["route"])
    print("Seat       :", f"{seat:02d}")
    print("Fare       : ₹", f"{fare:.2f}")
    print("Status     : CONFIRMED")

    print("=" * 45)





        
# ==================================================
# STEP 5: VIEW BOOKING
# ==================================================

def view_booking():

    print("\n========== VIEW BOOKING ==========")

    booking_id = input("Enter Booking ID: ").upper()

    # Check booking ID
    if booking_id not in bookings:

        print("Booking not found.")
        return

    booking = bookings[booking_id]

    print("\n========== BOOKING DETAILS ==========")

    print("Booking ID:", booking_id)
    print("Passenger Name:", booking["passenger_name"])
    print("Age:", booking["age"])
    print("Bus ID:", booking["bus_id"])
    print("Route:", booking["route"])
    print("Seat Number:", f"{booking['seat']:02d}")
    print("Fare: ₹", f"{booking['fare']:.2f}")
    print("Status:", booking["status"])


# ==================================================
# STEP 5: CANCEL TICKET
# ==================================================

def cancel_ticket():

    print("\n========== CANCEL TICKET ==========")

    booking_id = input("Enter Booking ID: ").upper()

    # Check booking
    if booking_id not in bookings:

        print("Booking not found.")
        return

    booking = bookings[booking_id]

    # Check if already cancelled
    if booking["status"] == "CANCELLED":

        print("This ticket is already cancelled.")
        return

    # Display booking information
    print("\nBooking Details")
    print("-------------------------")
    print("Booking ID:", booking_id)
    print("Passenger Name:", booking["passenger name"])
    print("Bus ID:", booking["bus_id"])
    print("Seat:", booking["seat"])
    print("Fare: ₹", booking["fare"])
    print("Status:", booking["status"])

    # Confirmation
    confirmation = input(
        "\nAre you sure you want to cancel this ticket? (yes/no): "
    ).lower()

    if confirmation != "yes":

        print("Cancellation stopped.")
        return

    # Get bus
    bus_id = booking["bus_id"]
    bus = buses[bus_id]

    # Release the seat
    seat = booking["seat"]

    if seat in buses[bus_id]["booked_seats"]:
        buses[bus_id]["booked_seats"].remove(seat)

    # Change booking status
    booking["status"] = "CANCELLED"

    print("\nTicket cancelled successfully.")
    print("Booking ID:", booking_id)
    print("Seat", seat, "has been released.")

# =========================================================
# STEP 6: SEARCH PASSENGER / BOOKING
# =========================================================

def search_booking():

    print("\n========== SEARCH PASSENGER / BOOKING ==========")

    search_value = input(
        "Enter Booking ID or Passenger Name: "
    ).strip()

    if search_value == "":
        print("Please enter something to search.")
        return

        
    # -----------------------------------------------------
    # Search by Booking ID
    # -----------------------------------------------------

    booking_id = search_value.upper()

    if booking_id in bookings:

        booking = bookings[booking_id]

        print("\n========== BOOKING FOUND ==========")

        print("Booking ID:", booking_id)
        print("Passenger Name:", booking["passenger_name"])
        print("Age:", booking["age"])
        print("Bus ID:", booking["bus_id"])
        print("Route:", booking["route"])
        print("Seat Number:", booking["seat"])
        print("Fare: ₹", booking["fare"])
        print("Status:", booking["status"])

        return

# -----------------------------------------------------
# Search by Passenger Name
# -----------------------------------------------------

    found = False

    for booking_id, booking in bookings.items():

        if search_value.lower() in booking["passenger_name"].lower():

            print("\n========== BOOKING FOUND ==========")

            print("Booking ID:", booking_id)
            print("Passenger Name:", booking["passenger_name"])
            print("Age:", booking["age"])
            print("Bus ID:", booking["bus_id"])
            print("Route:", booking["route"])
            print("Seat Number:", booking["seat"])
            print("Fare: ₹", booking["fare"])
            print("Status:", booking["status"])

            found = True
            
    if not found:

        print("\nNo passenger or booking found.")

     
# =========================================================
# STEP 6: SAVE DATA USING JSON
# =========================================================

def save_data():

    data = {
        "buses": buses,
        "bookings": bookings,
        "booking_counter": booking_counter
    }

    try:

        with open("booking_data.json", "w") as file:

            json.dump(data, file, indent=4)

        print("\nData saved successfully.")

    except Exception as e:

        print("Error while saving data:", e)


# =========================================================
# STEP 6: LOAD DATA USING JSON
# =========================================================

def load_data():

    global buses
    global bookings
    global booking_counter

    try:

        with open("booking_data.json", "r") as file:

            data = json.load(file)

        buses = data["buses"]
        bookings = data["bookings"]
        booking_counter = data["booking_counter"]

        print("Previous booking data loaded successfully.")

    except FileNotFoundError:

        print("No previous data found.")
        print("Starting with fresh data.")

    except json.JSONDecodeError:

        print("JSON file is empty or corrupted.")
        print("Starting with fresh data.")

    except Exception as e:

        print("Error while loading data:", e)


# ==========================================
#  MAIN MENU
# ==========================================

def display_menu():

    print("\n")
    print("=" * 45)
    print("          BUS RESERVATION SYSTEM")
    print("=" * 45)

    print("1. View Buses")
    print("2. View Available Seats")
    print("3. Book Ticket")
    print("4. Cancel Ticket")
    print("5. View Booking")
    print("6. Search Passenger / Booking")
    print("7. Save Data")
    print("8. Exit")

    print("=" * 45)


# ==========================================
# MAIN PROGRAM
# ==========================================

load_data()

while True:

    display_menu()

    choice = input("Enter your choice: ").strip()

    if choice == "1":

        view_buses()

    elif choice == "2":

        view_seats()

    elif choice == "3":

        book_ticket()

    elif choice == "4":

        cancel_ticket()

    elif choice == "5":

        view_booking()

    elif choice == "6":

        search_booking()

    elif choice == "7":

        save_data()

    elif choice == "8":

        # Save automatically before exiting
        save_data()

        print("\nThank you for using Bus Reservation System.")
        print("Goodbye!")

        break

    else:
        print("\nInvalid choice.")
        print("Please enter a number between 1 and 8.")
```
